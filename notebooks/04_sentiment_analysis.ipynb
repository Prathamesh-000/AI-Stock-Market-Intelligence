# 04 - Sentiment Analysis

Analyze sentiment in financial news using TextBlob and LLM extraction.


```python
import json
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob
from pathlib import Path

TICKERS = ['NVDA', 'AMD', 'AAPL', 'TSLA', 'MSFT']
NEWS_DIR = Path('../data/raw/news')
```

```python
# Basic TextBlob sentiment
results = []
for ticker in TICKERS:
    path = NEWS_DIR / f'{ticker}_news.json'
    if not path.exists():
        continue
    with open(path) as f:
        articles = json.load(f)
    for art in articles[:20]:
        blob = TextBlob(art.get('title', ''))
        results.append({
            'ticker': ticker,
            'headline': art.get('title', '')[:80],
            'polarity': blob.sentiment.polarity,
            'subjectivity': blob.sentiment.subjectivity
        })

df = pd.DataFrame(results)
print(df.groupby('ticker')['polarity'].describe())
```

```python
# Plot sentiment distribution
fig, ax = plt.subplots(figsize=(10, 5))
for ticker in TICKERS:
    subset = df[df['ticker'] == ticker]['polarity']
    if len(subset) > 0:
        subset.hist(ax=ax, alpha=0.6, bins=20, label=ticker)
ax.set_xlabel('Polarity Score')
ax.set_title('Sentiment Distribution by Ticker')
ax.legend()
plt.show()
```
