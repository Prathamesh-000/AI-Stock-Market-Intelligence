# 02 - Data Exploration

Exploratory data analysis on price and news data.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

TICKERS = ['NVDA', 'AMD', 'AAPL', 'TSLA', 'MSFT']
DATA_DIR = Path('../data/processed')
sns.set_theme(style='darkgrid')

# Load all datasets
dfs = {}
for ticker in TICKERS:
    path = DATA_DIR / f'{ticker}_final_dataset.csv'
    if path.exists():
        dfs[ticker] = pd.read_csv(path)
        print(f'{ticker}: {len(dfs[ticker])} rows, {len(dfs[ticker].columns)} columns')
```

```python
# Price distribution per ticker
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for i, ticker in enumerate(TICKERS):
    if ticker in dfs:
        df = dfs[ticker]
        if 'close' in df.columns:
            ax = axes[i//3][i%3]
            ax.plot(df['close'].values[-252:])
            ax.set_title(f'{ticker} - Last Year Price')
plt.tight_layout()
plt.show()
```

```python
# Sentiment score distribution
for ticker, df in dfs.items():
    if 'polarity_score' in df.columns:
        print(f'{ticker} sentiment stats:')
        print(df['polarity_score'].describe())
        print()
```
