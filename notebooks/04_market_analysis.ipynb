# 04 - Market Analysis

Analyze market microstructure: volume spikes, earnings impact windows, and sector correlations.


```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

TICKERS = ['NVDA', 'AMD', 'AAPL', 'TSLA', 'MSFT']

# Download 2 years of data
data = {}
for ticker in TICKERS:
    df = yf.download(ticker, period='2y', interval='1d', auto_adjust=True)
    df.columns = [c.lower() for c in df.columns]
    df['return'] = df['close'].pct_change()
    df['volume_zscore'] = (df['volume'] - df['volume'].rolling(20).mean()) / df['volume'].rolling(20).std()
    data[ticker] = df

print('Data loaded.')
```

```python
# Identify high-volume days (volume z-score > 2)
for ticker, df in data.items():
    high_vol = df[df['volume_zscore'] > 2.0]
    print(f'{ticker}: {len(high_vol)} high-volume events')
    print(f'  Avg return on high-vol days: {high_vol["return"].mean():.3%}')
    print(f'  Return std on high-vol days: {high_vol["return"].std():.3%}')
    print()
```

```python
# Return distribution
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for i, ticker in enumerate(TICKERS):
    ax = axes[i//3][i%3]
    data[ticker]['return'].dropna().hist(ax=ax, bins=50, color='steelblue', alpha=0.7)
    ax.axvline(0, color='red', linestyle='--')
    ax.set_title(f'{ticker} Daily Returns')
    ax.set_xlabel('Return')
plt.tight_layout()
plt.show()
```
