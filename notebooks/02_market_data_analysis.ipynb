# 02 - Market Data Analysis

Analyze OHLCV price data, compute technical indicators, and visualize market structure.


```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

TICKERS = ['NVDA', 'AMD', 'AAPL', 'TSLA', 'MSFT']

# Download 1 year of daily data
data = {}
for ticker in TICKERS:
    df = yf.download(ticker, period='1y', interval='1d', auto_adjust=True)
    df.columns = [c.lower() for c in df.columns]
    data[ticker] = df
    print(f'{ticker}: {len(df)} trading days')
```

```python
# Price comparison (normalized to 100)
fig, ax = plt.subplots(figsize=(12, 6))
for ticker, df in data.items():
    normalized = df['close'] / df['close'].iloc[0] * 100
    ax.plot(normalized.index, normalized.values, label=ticker)
ax.set_title('Relative Price Performance (Normalized to 100)')
ax.set_ylabel('Normalized Price')
ax.legend()
ax.grid(True, alpha=0.3)
plt.show()
```

```python
# Correlation matrix of returns
returns = pd.DataFrame({t: data[t]['close'].pct_change() for t in TICKERS}).dropna()
corr = returns.corr()

import seaborn as sns
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdYlGn', center=0, ax=ax)
ax.set_title('Return Correlation Matrix')
plt.show()
```
