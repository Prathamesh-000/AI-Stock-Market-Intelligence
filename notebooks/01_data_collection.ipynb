# 01 - Data Collection

This notebook covers downloading and saving raw financial data:
- OHLCV price data via yfinance
- News data from Yahoo Finance RSS
- Market indices (SPY, QQQ, VIX)


```python
import yfinance as yf
import pandas as pd
import json
import feedparser
from pathlib import Path

TICKERS = ['NVDA', 'AMD', 'AAPL', 'TSLA', 'MSFT']
DATA_DIR = Path('../data/raw')
```

```python
# Download price data
for ticker in TICKERS:
    df = yf.download(ticker, period='5y', interval='1d')
    df.to_csv(DATA_DIR / 'prices' / f'{ticker}_prices.csv')
    print(f'Downloaded {ticker}: {len(df)} rows')
```

```python
# Download news via RSS
for ticker in TICKERS:
    url = f'https://feeds.finance.yahoo.com/rss/2.0/headline?s={ticker}&region=US&lang=en-US'
    feed = feedparser.parse(url)
    articles = [{'title': e.title, 'summary': e.summary, 'published': e.published} for e in feed.entries]
    with open(DATA_DIR / 'news' / f'{ticker}_news.json', 'w') as f:
        json.dump(articles, f, indent=2)
    print(f'{ticker}: {len(articles)} articles')
```
